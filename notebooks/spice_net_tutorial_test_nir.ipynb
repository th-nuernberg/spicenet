# Spice Net
This is a quick guide on how to use the python spice net package.

Additionally, tests for its NIR exchange functionality.

````python
import numpy as np
import pandas as pd

import spice_net as spn
````

````python
df = pd.read_csv('data/test_data.csv')
df
````

## Initiation
First of all you need to create 2 self-organizing maps and 1 hebbian correlation matrix. The hardest part is figuring out the LRFs (= learning rate functions). You can find some implementations in the package, if you want to implement one for your self inherit the ```LearningRateFunction``` class. 

````python
som_size = 100
som_2_size = 100

som_1 = spn.SpiceNetSom(n_neurons=som_size,
                        value_range_start=df['init'].min(),
                        value_range_end=df['init'].max(),
                        lrf_tuning_curve=spn.ConstLRF(0.8),
                        lrf_interaction_kernel=spn.ConstLRF(0.8))
som_2 = spn.SpiceNetSom(n_neurons=som_2_size,
                        value_range_start=df['result'].min(),
                        value_range_end=df['result'].max(),
                        lrf_tuning_curve=spn.ConstLRF(0.8),
                        lrf_interaction_kernel=spn.ConstLRF(0.8))

correlation_matrix = spn.SpiceNetHcm(som_1, som_2, spn.ConstLRF(0.8), spn.ConstLRF(0.8))

spice_net = spn.SpiceNet(correlation_matrix)
````

## Fitting / Training
Try to play around with the ```epochs_on_batch``` and ```batch_size``` parameter. Depending on the amount of data you have available more epochs can improve or falsify your network. 

````python
spice_net.fit(df['init'].tolist(), df['result'].tolist(), 10, 100, print_output=True)
````

## Plotting
The package provides you with some plotting functions. Each of them starts with plot as name. Feel free to expirement with them.

````python
spn.plot_som(som_1, True)
spn.plot_som(som_2, True)
spn.plot_hcm(correlation_matrix)
````

## NIR Export

````python
# Compare result over 100 samples
errors = []

for i in range(100):
    test_data = df.sample()
    inti_test_value = test_data['init'].iloc[0]
    result_test_value = test_data['result'].iloc[0]
    try:
        predicted = spice_net.decode(inti_test_value)
    except Exception as e:
        print("Error in original spicenet, continuing..")
        continue
    predicted_imported = imported_spicenet.decode(inti_test_value)
    predicted_imported_no_lrf = imported_spicenet_no_lrf.decode(inti_test_value)
    assert predicted == predicted_imported
    assert predicted == predicted_imported_no_lrf
    errors.append(abs(predicted - result_test_value) / 2 * 100)
    print(f'Error: {errors[-1]:.6f},\t Predicted: {predicted:.6f},\t Predicted Imported: {predicted_imported:.6f},\t Predicted Imported No LRF: {predicted_imported_no_lrf:.6f},\t Actual: {result_test_value:.6f}')

print(f'Mean Error in % {np.mean(np.array(errors))}')
print(f'Median Error in % {np.median(np.array(errors))}')
    
````

# Test export metadata for all learning rate functions
We will create a newly trained spicenet for every LRF available and then export and import it and make sure it has the correct parameters

````python
import inspect
import spice_net.learning_rate_functions as lrf
import random
import string

classes = {name: obj for name, obj in inspect.getmembers(lrf, inspect.isclass)}
classes.pop("LearningRateFunction") # Pop Base Class
classes
````

````python
# We want to get random instantiation parameters for each class. We don't care if they would be correct in a real problem as we only want to test their transfer
def random_value(param_type):
    """Generate a random value based on the parameter type."""
    if param_type is int:
        return random.randint(1, 10000)
    elif param_type is float:
        return random.uniform(0.0, 1.0)
    else:
        return None  # Handle unknown types


args_for_class = {}

# For each class
for name, classtype in classes.items():
    # Get signature of the constructor
    signature = inspect.signature(classtype.__init__)
    # Get parameters of the constructor without self
    parameters = dict(signature.parameters)
    parameters.pop("self")
    
    # Get arguments for instantiation by type without default values being changed
    args_for_class[name] = {name: random_value(param.annotation) for name, param in parameters.items() if param.default == inspect.Parameter.empty}
print(args_for_class)
````

````python
# Define function to compare two spicenets
def compare_spn(spice_net1, spice_net2):
    # Compare spicenet to real spicenet
    assert len(spice_net2.som_1.neurons) == len(spice_net1.som_1.neurons)
    assert len(spice_net2.som_2.neurons) == len(spice_net1.som_2.neurons)

    for i in range(som_size):
        assert spice_net2.som_1.neurons[i].tuning_curve_width == spice_net1.som_1.neurons[i].tuning_curve_width
        assert spice_net2.som_1.neurons[i].preferred_value == spice_net1.som_1.neurons[i].preferred_value

    for i in range(som_2_size):
        assert spice_net2.som_2.neurons[i].tuning_curve_width == spice_net1.som_2.neurons[i].tuning_curve_width
        assert spice_net2.som_2.neurons[i].preferred_value == spice_net1.som_2.neurons[i].preferred_value

    assert spice_net2.get_correlation_matrix().weights.shape == spice_net1.get_correlation_matrix().weights.shape
    assert np.all(spice_net2.get_correlation_matrix().weights == spice_net1.get_correlation_matrix().weights)

    assert np.all(spice_net2.get_correlation_matrix().activation_bar_vector_1 == spice_net1.get_correlation_matrix().activation_bar_vector_1)
    assert np.all(spice_net2.get_correlation_matrix().activation_bar_vector_2 == spice_net1.get_correlation_matrix().activation_bar_vector_2)
    assert spice_net2.get_correlation_matrix().get_iteration() == spice_net1.get_correlation_matrix().get_iteration()
    assert spice_net2.get_correlation_matrix().get_trust_of_new_lrf().get_parameters() == spice_net1.get_correlation_matrix().get_trust_of_new_lrf().get_parameters()
    assert spice_net2.get_correlation_matrix().get_weights_lrf().get_parameters() == spice_net1.get_correlation_matrix().get_weights_lrf().get_parameters()
    assert spice_net2.get_correlation_matrix().get_trust_of_new_lrf().__class__.__name__ == spice_net1.get_correlation_matrix().get_trust_of_new_lrf().__class__.__name__
    assert spice_net2.get_correlation_matrix().get_weights_lrf().__class__.__name__ == spice_net1.get_correlation_matrix().get_weights_lrf().__class__.__name__

    assert spice_net2.som_1.get_iteration() == spice_net1.som_1.get_iteration()
    assert spice_net2.som_2.get_iteration() == spice_net1.som_2.get_iteration()

    assert spice_net2.som_1.get_lrf_tuning_curve().get_parameters() == spice_net1.som_1.get_lrf_tuning_curve().get_parameters()
    assert spice_net2.som_1.get_lrf_interaction_kernel().get_parameters() == spice_net1.som_1.get_lrf_interaction_kernel().get_parameters()

    assert spice_net2.som_2.get_lrf_tuning_curve().get_parameters() == spice_net1.som_2.get_lrf_tuning_curve().get_parameters()
    assert spice_net2.som_2.get_lrf_interaction_kernel().get_parameters() == spice_net1.som_2.get_lrf_interaction_kernel().get_parameters()
````

````python
# Train a spicenet with each learning rate function and the random args for instantiation, then export it, import it and validate all parameters
for name, classtype in classes.items():
    print(f"Training with {name}({args_for_class[name]})")
    som_1 = spn.SpiceNetSom(n_neurons=som_size,
                            value_range_start=df['init'].min(),
                            value_range_end=df['init'].max(),
                            lrf_tuning_curve=classtype(**args_for_class[name]),
                            lrf_interaction_kernel=classtype(**args_for_class[name]))
    som_2 = spn.SpiceNetSom(n_neurons=som_2_size,
                            value_range_start=df['result'].min(),
                            value_range_end=df['result'].max(),
                            lrf_tuning_curve=classtype(**args_for_class[name]),
                            lrf_interaction_kernel=classtype(**args_for_class[name]))

    correlation_matrix = spn.SpiceNetHcm(som_1, som_2, classtype(**args_for_class[name]), classtype(**args_for_class[name]))

    spice_net = spn.SpiceNet(correlation_matrix)
    spice_net.fit(df['init'].tolist(), df['result'].tolist(), 10, 100, print_output=True)
    
    # Export trained
    try:
        exported_spice_net = to_nir(spice_net)
    except AssertionError as e:
        print(f"Error: {e}")
        print("Unsuccessful export due to invalid parameters.. not the fault of lrf export \n\n")
        
        # Check neurons
        has_zero_or_less = False
        has_nan = False
        for i in range(som_size):
            if spice_net.som_1.neurons[i].tuning_curve_width <= 0 or spice_net.som_2.neurons[i].tuning_curve_width <= 0:
                has_zero_or_less = True
            if np.isnan(spice_net.som_1.neurons[i].tuning_curve_width) or np.isnan(spice_net.som_2.neurons[i].tuning_curve_width):
                has_nan = True
        if has_zero_or_less:
            print("Reason: Neurons have values with STD that is 0 or less. This is not allowed. Correct.\n\n")
        if has_nan:
            print("Reason: Neurons have values with STD that is NaN. So the error is correct.\n\n")
        
        continue
    
    # Import trained without learning rate functions
    imported_spice_net_without_learn = from_nir(exported_spice_net)
    try:
        compare_spn(spice_net, imported_spice_net_without_learn)
        print("Comparison successful\n\n")
    except AssertionError as e:
        print(f"Error: {e}")
        print("Unsuccessful comparison \n\n")
   
````

We see success wherever training was successful. When training failed we automatically analyse the problem and see that it results from either not being a number (NaN) or being below 0. This is both valid reasons for NIR to not accept the export and as such is exactl what we want.

However, to make sure it does we do the same thing but without actual fitting to see if initial values would be transfered since they should always be valid for all Learning Rate Functions.

````python
# Train a spicenet with each learning rate function and the random args for instantiation, then export it, import it and validate all parameters
for name, classtype in classes.items():
    print(f"Training with {name}({args_for_class[name]})")
    som_1 = spn.SpiceNetSom(n_neurons=som_size,
                            value_range_start=df['init'].min(),
                            value_range_end=df['init'].max(),
                            lrf_tuning_curve=classtype(**args_for_class[name]),
                            lrf_interaction_kernel=classtype(**args_for_class[name]))
    som_2 = spn.SpiceNetSom(n_neurons=som_2_size,
                            value_range_start=df['result'].min(),
                            value_range_end=df['result'].max(),
                            lrf_tuning_curve=classtype(**args_for_class[name]),
                            lrf_interaction_kernel=classtype(**args_for_class[name]))

    correlation_matrix = spn.SpiceNetHcm(som_1, som_2, classtype(**args_for_class[name]), classtype(**args_for_class[name]))

    spice_net = spn.SpiceNet(correlation_matrix)
    
    # Export constructed but untrained
    try:
        exported_spice_net = to_nir(spice_net)
    except AssertionError as e:
        print(f"Error: {e}")
        print("Unsuccessful export due to invalid parameters.. not the fault of lrf export \n\n")
        
        # Check neurons
        has_zero_or_less = False
        has_nan = False
        for i in range(som_size):
            if spice_net.som_1.neurons[i].tuning_curve_width <= 0 or spice_net.som_2.neurons[i].tuning_curve_width <= 0:
                has_zero_or_less = True
            if np.isnan(spice_net.som_1.neurons[i].tuning_curve_width.nan) or np.isnan(spice_net.som_2.neurons[i].tuning_curve_width):
                has_nan = True
        if has_zero_or_less:
            print("Reason: Neurons have values with STD that is 0 or less. This is not allowed. Correct.")
        if has_nan:
            print("Reason: Neurons have values with STD that is NaN. So the error is correct.")
        
        continue
    
    # Import trained without learning rate functions
    imported_spice_net_without_learn = from_nir(exported_spice_net)
    try:
        compare_spn(spice_net, imported_spice_net_without_learn)
        print("Comparison successful\n\n")
    except AssertionError as e:
        print(f"Error: {e}")
        print("Unsuccessful comparison \n\n")
   
````
